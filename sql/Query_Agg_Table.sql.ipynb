##Using SQL output the following aggregates
####Profit by Year
####Profit by Year + Product Category
####Profit by Customer
####Profit by Customer + Year


```python
%sql

-- Profit by Year
WITH profit_by_year AS (
    SELECT 
        Year,
        SUM(Total_Profit) AS Profit_By_Year
    FROM gold.profit_aggregates
    GROUP BY Year
),

-- Profit by Year + Product Category
profit_by_year_category AS (
    SELECT 
        Year,
        Category,
        SUM(Total_Profit) AS Profit_By_Year_Category
    FROM gold.profit_aggregates
    GROUP BY Year, Category
),

-- Profit by Customer
profit_by_customer AS (
    SELECT 
        Customer_ID,
        SUM(Total_Profit) AS Profit_By_Customer
    FROM gold.profit_aggregates
    GROUP BY Customer_ID
),

-- Profit by Customer + Year
profit_by_customer_year AS (
    SELECT 
        Customer_ID,
        Year,
        SUM(Total_Profit) AS Profit_By_Customer_Year
    FROM gold.profit_aggregates
    GROUP BY Customer_ID, Year
)

SELECT * FROM profit_by_year;
-- SELECT * FROM profit_by_year_category;
-- SELECT * FROM profit_by_customer;
-- SELECT * FROM profit_by_customer_year;
```
