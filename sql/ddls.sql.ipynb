```python
%sql
CREATE TABLE IF NOT EXISTS silver.order_details (
    Order_ID STRING,
    Order_Date STRING,
    Customer_ID STRING,
    Customer_Name STRING,
    Country STRING,
    Category STRING,
    Sub_Category STRING,
    Profit DECIMAL(10,2),

    -- Audit columns
    created_at TIMESTAMP,
    modified_at TIMESTAMP,
    created_job_id STRING,
    created_run_id STRING,
    modified_job_id STRING,
    modified_run_id STRING
)
USING DELTA
PARTITIONED BY (Country, Category)  -- recommended partitioning
LOCATION 'dbfs:/mnt/silver/order_details';


CREATE TABLE IF NOT EXISTS gold.profit_aggregates (
    Year INT,
    Category STRING,
    Sub_Category STRING,
    Customer_ID STRING,
    Total_Profit DOUBLE,
    created_at TIMESTAMP,
    modified_at TIMESTAMP,
    created_job_id STRING,
    created_run_id STRING,
    modified_job_id STRING,
    modified_run_id STRING
)
USING delta
PARTITIONED BY (Year, Category);

CREATE TABLE IF NOT EXISTS error.profit_aggregates_bad_records (
    Year INT,
    Category STRING,
    Sub_Category STRING,
    Customer_ID STRING,
    Total_Profit DOUBLE,
    created_at TIMESTAMP,
    modified_at TIMESTAMP,
    created_job_id STRING,
    created_run_id STRING,
    modified_job_id STRING,
    modified_run_id STRING
)
USING delta
PARTITIONED BY (Year, Category);









```
